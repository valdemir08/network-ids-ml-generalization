```python
from pathlib import Path

import pandas as pd

DATA_DIR = Path("../datasets/final/ml")

df_cic = pd.read_parquet(
    DATA_DIR / "cicids2017_sampled.parquet"
)

df_unsw = pd.read_parquet(
    DATA_DIR / "unsw_nb15_sampled.parquet"
)

df_iot = pd.read_parquet(
    DATA_DIR / "iot23_sampled.parquet"
)
```

```python
display(df_cic.shape)
```

```python
display(df_unsw.shape)
```

```python
display(df_iot.shape)
```

```python
display(df_cic["label_binary"].value_counts())
display(df_unsw["label_binary"].value_counts())
display(df_iot["label_binary"].value_counts())
```

```python
columns_to_check = [
    "label",
    "label_binary",
    "match_status",
]

display(df_cic[columns_to_check].isna().sum())
display(df_unsw[columns_to_check].isna().sum())
display(df_iot[columns_to_check].isna().sum())
```

```python
from sklearn.model_selection import train_test_split
```

```python
df_cic_train, df_cic_test = train_test_split(
    df_cic,
    test_size=0.30,
    stratify=df_cic["label_binary"],
    random_state=1,
)
```

```python
df_unsw_train, df_unsw_test = train_test_split(
    df_unsw,
    test_size=0.30,
    stratify=df_unsw["label_binary"],
    random_state=1,
)
```

```python
df_iot_train, df_iot_test = train_test_split(
    df_iot,
    test_size=0.30,
    stratify=df_iot["label_binary"],
    random_state=1,
)
```

```python


df_cic_train.to_parquet(
    DATA_DIR / "cicids2017_train.parquet",
    index=False,
)

df_cic_test.to_parquet(
    DATA_DIR / "cicids2017_test.parquet",
    index=False,
)

df_unsw_train.to_parquet(
    DATA_DIR / "unsw_nb15_train.parquet",
    index=False,
)

df_unsw_test.to_parquet(
    DATA_DIR / "unsw_nb15_test.parquet",
    index=False,
)

df_iot_train.to_parquet(
    DATA_DIR / "iot23_train.parquet",
    index=False,
)

df_iot_test.to_parquet(
    DATA_DIR / "iot23_test.parquet",
    index=False,
)
```
